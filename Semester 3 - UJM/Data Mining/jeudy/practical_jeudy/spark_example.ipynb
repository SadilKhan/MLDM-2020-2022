```python
from pyspark import SparkContext
sc = SparkContext("local", "example")
```

```python
# create a rdd from a python list
rdd1 = sc.parallelize( [13, 2, 33, 24, 5, 7, 234, 23, 42, 12, 2, 33, 55] )
rdd1.collect()
```

```python
# example of map
rdd2 = rdd1.map(lambda x: 1+x)
rdd2.take(3)
```

```python
# another example of map
rdd2 = rdd1.map(lambda x: (x,2*x))
rdd2.collect()
```

```python
# example of flatmap (with the same function as above)
# the result is different than above
rdd2 = rdd1.flatMap(lambda x: (x,2*x))
rdd2.collect()
```

```python
# example of filter
rddb = sc.parallelize( [("banana",3),("apple",2),("cherry",12),("pear",5),("chestnut",2)] )

def f(x):
    return x[1] > 4

rdd3=rddb.filter(f)
rdd3.collect()
```

```python
# compute the max price for each product (see exercise 1 in the exercise sheet)
pasta_txt = sc.textFile("data/pasta.txt")
pasta_txt.take(5)
```

```python
def split_pasta(line):
    cpny, prod, price = line.split()
    return (cpny, prod, int(price))

pasta = pasta_txt.map(split_pasta)
pasta.collect()
```

```python
prod_price = pasta.map(lambda p: (p[1], p[2]) )
prod_price.collect()
```

```python
max_price = prod_price.reduceByKey(max)
max_price.collect()
```

```python
# another example of filter
prod2_price = prod_price.filter(lambda x: x[0]=='bavette' or x[0]=='penne')
prod2_price.collect()
```

```python
# Word count :
# input is a text file
# output is a set of pairs (word, frequency) for each word of the text
text=sc.textFile("data/lotr.txt")
text.take(3)
```

```python
wordlist=text.flatMap(lambda line: line.lower().split())
wordlist.take(100)
```

```python
pairs = wordlist.map(lambda x: (x,1))
pairs.take(5)
```

```python
word_frequencies = pairs.reduceByKey(lambda x,y: x+y).map(lambda x:(x[1],x[0])).sortByKey(ascending = False)
word_frequencies.take(10)
```

```python
# with sortBy
word_frequencies = pairs.reduceByKey(lambda x,y: x+y).sortBy(lambda x: x[1], ascending = False)
word_frequencies.take(10)
```

```python
# one doc for each row of text
# produce (did, doc) pairs where doc is a row from text
docs = sc.parallelize(enumerate(text.collect()))
docs.take(3)
```

```python
def split_words(x):
    did, text = x
    return [((did,word),1) for word in text.split() ]

words = docs.flatMap(split_words)
words.takeSample(False, 10)
```

```python
res = words.reduceByKey(lambda x,y: x+y)
res.takeSample(False, 10)
```
